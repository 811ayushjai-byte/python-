```python
#List Operations
#Append()-Adds one element at the end

d=[1,2,3,4]
print("before Append",d)
d.append("five")
print("After Append",d)
```

```python
#Extend()- Adds multiple elements

e=[1,2,3,4]
print("before extend", d)
e.extend(["five","six", "true" , 5])
print("after extend", e)
```

```python
#Insert()-Adds element at a specific position

f=["java","python","js","html"]
print(f)
```

```python
f.insert(0,"ml")
```

```python
f
```

```python
#Remove-removes the first occurrence of a value

a=[10,20,30,20]
a.remove(20)
print(a)

```

```python
#pop()-Removes and returns an element

a=[10,20,30]
x=a.pop()
print(x)
print(a)
```

```python
#clear()-Removes all elements

a=[1,2,3,]
a.clear()
print(a)

```

```python
#Index()-Returns the index of an element
a=[10,20,30,40]
print(a.index(30))
```

```python
#Count-counts how many times an element occurs
a=[10,20,20,30,20]
print(a.count(20))
```

```python
#Sort- sorts the list
a=[40,10,30,20]
a.sort()
print(a)
```

```python
#Descending order:

a=[40,10,30,20]
a.sort(reverse=True)
print(a)
```

```python
#Reverse()-reverse the list
a=[1,2,3,4]
a.reverse()
print(a)
```

```python
#Copy()-Creates a shallow copy

a=[1,2,3]
b=a.copy()
print(b)
```

```python
#Operators on lists
#Concatenation+

a=[1,2]
b=[3,4]
print(a+b)
```

```python
#Repetition*

a=[1,2]
print(a*3)
```

```python
#Membership in

a=[10,20,30]
print(20 in a)
print(50 in a)
```

```python
# not in

print(50 not in a)
```

```python
#Comparison

print([1,2]==[1,2])
print([1,2]!=[1,3])
```

```python
#Useful Built-in Functions

a=[10,20,30,40]

print(len(a))
print(min(a))
print(max(a))
print(sum(a))
```
