```python
# And Operator

is_aadhaar = input("Do you have Aadhaar ")
is_pan = input("Do you have PAN ")
age = 20

if is_aadhaar == "yes" and is_pan == "yes" and age >= 18:
    print("Eligible")
else:
    print("Not eligible")
```

```python
# Or Operator

is_aadhaar = input("Do you have Aadhaar")
is_pan = input("Do you have PAN")
age = 20
if is_aadhaar == "yes" or is_pan == "yes" or age < 18:
    print("eligible")
else:
    print("not eligible")
```

```python
# Not Operator

is_aadhaar = input("Do you have Aadhaar")
is_pan = input("Do you have PAN")
age = 20
if not (is_aadhaar != "yes" or is_pan != "yes" or age < 18):
    print("Eligible")
else:
    print("Not eligible")
```
